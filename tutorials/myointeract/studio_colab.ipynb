# MyoInteract Studio on Colab

Studio is the guided web app of MyoInteract: set up a body, a task, sensing and rewards in front of a 3D scene, train on the GPU of this Colab session, watch the policy live and evaluate it.

[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/MyoInteract/MyoInteract/blob/main/tutorials/myointeract/studio_colab.ipynb)

1. Use a GPU runtime (*Runtime → Change runtime type → T4 GPU*).
2. Run the two cells below. The second one prints the link **Open MyoInteract Studio**, which opens Studio in a new tab through Colab's port proxy (only your Google account gets in).

A free T4 is about five times slower than an RTX 5090: the pointing example takes about 20 minutes. Runs are lost when the session ends unless you store them on Drive (see the second cell).

```python
# Install MyoInteract (a few minutes).
import importlib.util, subprocess, sys

if importlib.util.find_spec("myoInteract") is None or importlib.util.find_spec("mjlab") is None:
    done = subprocess.run([sys.executable, "-m", "pip", "install", "-q", "myoInteract[colab] @ git+https://github.com/MyoInteract/MyoInteract.git@v1.0.0"], capture_output=True, text=True)
    if done.returncode != 0:
        print(done.stderr[-1500:])
        raise RuntimeError("MyoInteract could not be installed.")
    if "google.colab" in sys.modules:
        print("Installed. If the next cell fails to import, restart the runtime (Runtime -> Restart session) and run again from here.")
```

```python
from myoInteract.studio.colab import launch

# To keep runs and presets after the session, mount Drive and pass a folder on it:
#   from google.colab import drive; drive.mount("/content/drive")
#   studio = launch(root="/content/drive/MyDrive/myoInteract")
studio = launch()
```

Stop Studio with `studio.terminate()`. Its log is `studio_8000.log`; training runs and their checkpoints are in `myoInteract_data/logs/` (or in the folder passed as `root`).
